##Task 1
Mulailah memuat semua data yang ada ke sistem.

```python
import pandas as pd

# 1) Muat data dari kedua sumber
df_periode1 = pd.read_csv('/content/Data Siswa Periode 1.csv')
df_periode2 = pd.read_csv('/content/Data Siswa Periode 2.csv')

# 2) Lihat atribut apa saja yang ada dari kedua data tersebut
print("=== Atribut Data Periode 1 ===")
print(f"Ukuran data: {df_periode1.shape}")
print(df_periode1.info())

print("\n=== Atribut Data Periode 2 ===")
print(f"Ukuran data: {df_periode2.shape}")
print(df_periode2.info())

print("\nBeberapa baris pertama Periode 1:")
display(df_periode1.head())

print("\nBeberapa baris pertama Periode 2:")
display(df_periode2.head())
```

##Task 2
Datanya sudah termuat, fitur nya hampir semuanya sama, tetapi ada yang berbeda sehingga perlu dipetakan

```python
# 1) Identifikasi fitur serupa, lakukan penamaan ulang fitur serupa
# Menyamakan nama kolom dari df_periode2 agar sesuai dengan df_periode1
df_periode2 = df_periode2.rename(columns={
    'Kelas Siang/Malam': 'Kelas Reguler/Malam',
    'Umur saat mendaftar': 'Usia saat mendaftar'
})

# Verifikasi nama kolom setelah diubah
print("Kolom Periode 1:")
print(df_periode1.columns.tolist())
print("\nKolom Periode 2:")
print(df_periode2.columns.tolist())
```

##Task 3
Butuh penyatuan data, tapi ternyata ada kodefikasi atribut fitur yang berbeda, harus di samakan

```python
# 1) Lakukan penyamaan kodefikasi atribut
# Mari kita periksa nilai unik pada beberapa kolom kategorikal penting yang berpotensi memiliki perbedaan kodefikasi
print("=== Unik Kelas Reguler/Malam ===")
print("Periode 1:", df_periode1['Kelas Reguler/Malam'].unique())
print("Periode 2:", df_periode2['Kelas Reguler/Malam'].unique())

print("\n=== Unik Target ===")
print("Periode 1:", df_periode1['Target'].unique())
print("Periode 2:", df_periode2['Target'].unique())

# Berdasarkan pemeriksaan nilai unik:
# Pada Periode 1 terdapat 'Reguler' dan 'Malam'. Pada Periode 2 terdapat 'Siang' dan 'Malam'.
# Kita samakan 'Siang' pada Periode 2 menjadi 'Reguler' agar konsisten.
df_periode2['Kelas Reguler/Malam'] = df_periode2['Kelas Reguler/Malam'].replace({'Siang': 'Reguler'})

# 2) Lakukan penyatuan data
df_gabungan = pd.concat([df_periode1, df_periode2], ignore_index=True)

print("\n=== Hasil Penyatuan Data ===")
print(f"Ukuran data gabungan: {df_gabungan.shape}")
print(df_gabungan['Kelas Reguler/Malam'].value_counts(dropna=False))
```

##Task 4
Sekarang kita lakukan analisis atribut untuk setiap fitur nya, mulai dengan penanganan missing/null value

```python
# 1)	Identifikasi dan tangani missing/null value
print("=== Jumlah Missing Values per Kolom ===")
missing_info = df_gabungan.isnull().sum()
print(missing_info[missing_info > 0])

# Menangani missing value pada kolom 'Kelas Reguler/Malam' dengan menggunakan modus
if 'Kelas Reguler/Malam' in df_gabungan.columns:
    modus_kelas = df_gabungan['Kelas Reguler/Malam'].mode()[0]
    df_gabungan['Kelas Reguler/Malam'] = df_gabungan['Kelas Reguler/Malam'].fillna(modus_kelas)

# Menangani missing value pada kolom lain (jika ada)
# Untuk kolom numerik jika ada yang kosong akan kita isi dengan median, dan kategorikal dengan modus
for col in df_gabungan.columns:
    if df_gabungan[col].isnull().any():
        if df_gabungan[col].dtype == 'object':
            df_gabungan[col] = df_gabungan[col].fillna(df_gabungan[col].mode()[0])
        else:
            df_gabungan[col] = df_gabungan[col].fillna(df_gabungan[col].median())

print("\n=== Verifikasi Missing Values Setelah Penanganan ===")
print(f"Total missing values: {df_gabungan.isnull().sum().sum()}")
```

##Task 5
Untuk bisa melihat bentuk dari data, kita perlu menghitung histogram dari setiap fitur. Dari situ kemudian kita bisa lanjutkan dengan penanganan outlier.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Tentukan kolom numerik dan kategorikal
numerik_cols = ['Nilai SMA', 'Nilai Ujian Masuk', 'Usia saat mendaftar', 'IP Semester 1', 'IP Semester 2']
kategorikal_cols = [col for col in df_gabungan.columns if col not in numerik_cols and col != 'ID']

# 1) Hitung histogram setiap fitur, tampilkan dalam bentuk grafik batang/distribusi
plt.figure(figsize=(18, 12))
for i, col in enumerate(numerik_cols, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df_gabungan[col], kde=True, color='skyblue')
    plt.title(f'Distribusi {col}')
plt.tight_layout()
plt.show()

# 2) Identifikasi outlier dengan boxplot
plt.figure(figsize=(15, 6))
sns.boxplot(data=df_gabungan[numerik_cols])
plt.title('Identifikasi Outlier pada Fitur Numerik')
plt.xticks(rotation=15)
plt.show()

# 3) Hapus outlier dari data menggunakan metode IQR
print(f"Ukuran data sebelum menghapus outlier: {df_gabungan.shape}")

for col in numerik_cols:
    Q1 = df_gabungan[col].quantile(0.25)
    Q3 = df_gabungan[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter data yang berada di dalam range non-outlier
    df_gabungan = df_gabungan[(df_gabungan[col] >= lower_bound) & (df_gabungan[col] <= upper_bound)]

print(f"Ukuran data setelah menghapus outlier: {df_gabungan.shape}")
```
